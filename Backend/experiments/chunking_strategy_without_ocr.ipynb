```python
import fitz
file_path = r"C:\Users\SumeetMaheshwari\Desktop\workspace\doc-comparator\sample\Output 1.pdf"
```

```python
doc = fitz.open(file_path)
```

```python
emp_text = ""
for i, page in enumerate(doc):
    text = doc.get_page_text(pno=i)
    emp_text += text
```
